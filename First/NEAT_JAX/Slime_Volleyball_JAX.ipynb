```python
# --- Global NEAT Parameters ---
POPULATION_SIZE = 100
NUM_GENERATIONS =  100
COMPATIBILITY_THRESHOLD = 0.5
C1 = 1.0  # Excess genes coefficient
C2 = 1.0  # Disjoint genes coefficient
C3 = 0.4  # Weight difference coefficient
MUTATION_RATE_ADD_CONNECTION = 0.1
MUTATION_RATE_ADD_NODE = 0.1
MUTATION_RATE_PERTURB_WEIGHT = 1
MUTATION_RATE_RESET_WEIGHT = 0.1
WEIGHT_PERTURB_SCALE = 0.5
ELITISM_PERCENTAGE = 0.1
NUM_INPUTS = 12  # SlimeVolley observations
NUM_OUTPUTS = 3  # SlimeVolley actions (forward, backward, jump)
```

##Definitions

```python
# --- Activation Function ---
def sigmoid(x):
    return 1 / (1 + np.exp(-np.clip(x, -10, 10))) # Clip to prevent overflow

# --- Gene Classes ---
class NodeGene:
    def __init__(self, node_id, node_type):
        self.node_id = node_id  # Unique ID for the node
        self.node_type = node_type  # 'input', 'hidden', 'output'
        self.bias = np.random.uniform(-1, 1)

class ConnectionGene:
    def __init__(self, in_node_id, out_node_id, weight, innovation_num, enabled=True):
        self.in_node_id = in_node_id
        self.out_node_id = out_node_id
        self.weight = weight
        self.innovation_num = innovation_num  # Global innovation ID
        self.enabled = enabled

# --- Genome Class ---
class Genome:
    def __init__(self, genome_id):
        self.genome_id = genome_id
        self.node_genes = {}  # {node_id: NodeGene}
        self.connection_genes = {}  # {innovation_num: ConnectionGene}
        self.fitness = 0.0
        self.species_id = -1

    def calculate_output(self, inputs):
        # Create a mapping from node_id to its calculated value
        node_values = {}

        # Initialize input node values
        for i, input_val in enumerate(inputs):
            node_id = i
            node_values[node_id] = input_val

        # Add bias to input nodes (optional, often handled by a bias node)
        # For simplicity, we'll just have bias on hidden/output nodes for now.

        # Topological sort or repeated passes to ensure all inputs to a node are computed
        # For strictly feed-forward, we can iterate from input to output layers.
        # This is a simplified approach, a more robust NEAT would build the graph
        # and traverse it topologically.

        # Collect active connections
        active_connections = [
            conn for conn in self.connection_genes.values() if conn.enabled
        ]

        # Sort connections by output node ID to process layers
        # This is a heuristic for feed-forward, proper topological sort is better
        # For simplicity, we'll just iterate a few times, assuming small depth.
        # A more robust solution involves building the graph and performing a real topological sort.
        # For now, let's just make multiple passes, assuming a shallow network.
        # A better approach would be to build the actual graph and use networkx or similar for topological sort.

        # Max passes to ensure all nodes are calculated (assumes limited depth)
        for _ in range(self.get_max_depth()): # Max depth based on connections
            for conn in active_connections:
                if conn.in_node_id in node_values:
                    # Calculate input to the output node
                    incoming_sum = node_values.get(conn.out_node_id, 0.0) + conn.weight * node_values[conn.in_node_id]
                    node_values[conn.out_node_id] = incoming_sum # Sum up inputs

            # Apply activation and bias after all inputs to a node are summed
            for node_id, node_gene in self.node_genes.items():
                if node_id >= NUM_INPUTS and node_id in node_values: # Only apply activation/bias to hidden/output
                    node_values[node_id] = sigmoid(node_values[node_id] + node_gene.bias)


        # Get output values
        outputs = []
        for i in range(NUM_INPUTS + self.get_num_hidden_nodes(), NUM_INPUTS + self.get_num_hidden_nodes() + NUM_OUTPUTS):
            outputs.append(node_values.get(i, 0.0)) # Default to 0 if node not reached
        return np.array(outputs)


    def get_max_depth(self):
        # A very simplified way to estimate depth for feedforward
        # In a real NEAT, you'd build a graph and calculate depth
        max_depth = 2 # Input -> Output minimum
        node_levels = {i: 0 for i in range(NUM_INPUTS)} # Input nodes are level 0

        # Keep track of nodes whose levels have been determined
        processed_nodes = set(range(NUM_INPUTS))

        # Iteratively assign levels to nodes
        changed = True
        while changed:
            changed = False
            for conn in self.connection_genes.values():
                if conn.enabled and conn.in_node_id in processed_nodes and conn.out_node_id not in processed_nodes:
                    node_levels[conn.out_node_id] = node_levels[conn.in_node_id] + 1
                    processed_nodes.add(conn.out_node_id)
                    max_depth = max(max_depth, node_levels[conn.out_node_id] + 1) # +1 for layer after node
                    changed = True
        return max_depth


    def get_num_hidden_nodes(self):
        count = 0
        for node_id, node_gene in self.node_genes.items():
            if node_gene.node_type == 'hidden':
                count += 1
        return count
```

```python
# --- Innovation Tracker ---
class InnovationTracker:
    def __init__(self):
        self.innovation_history = {}  # (in_node, out_node) -> innovation_num
        self.next_innovation_num = 0
        self.next_node_id = NUM_INPUTS + NUM_OUTPUTS # Start node IDs after fixed input/output

    def get_innovation_num(self, in_node, out_node):
        if (in_node, out_node) not in self.innovation_history:
            self.innovation_history[(in_node, out_node)] = self.next_innovation_num
            self.next_innovation_num += 1
        return self.innovation_history[(in_node, out_node)]

    def get_next_node_id(self):
        self.next_node_id += 1
        return self.next_node_id - 1
```

```python
# --- Species Class ---
class Species:
    def __init__(self, species_id, representative_genome):
        self.species_id = species_id
        self.representative = representative_genome
        self.genomes = []
        self.average_fitness = 0.0

    def add_genome(self, genome):
        self.genomes.append(genome)
        genome.species_id = self.species_id

    def calculate_average_fitness(self):
        if not self.genomes:
            self.average_fitness = 0.0
        else:
            self.average_fitness = sum(g.fitness for g in self.genomes) / len(self.genomes)

    def sort_genomes(self):
        self.genomes.sort(key=lambda g: g.fitness, reverse=True)

    def select_representative(self):
        if self.genomes:
            self.representative = random.choice(self.genomes) # Could also be the fittest

```

```python
# --- NEAT Algorithm Class ---
class NEAT:
    def __init__(self, input_size, output_size):
        self.input_size = input_size
        self.output_size = output_size
        self.population = []
        self.species = []
        self.innovation_tracker = InnovationTracker()
        self.generation = 0
        self.next_genome_id = 0
        self.next_species_id = 0

    def create_initial_population(self):
        for _ in range(POPULATION_SIZE):
            genome = self._create_minimal_genome()
            self.population.append(genome)
            self.next_genome_id += 1

    def _create_minimal_genome(self):
        genome = Genome(self.next_genome_id)

        # Input nodes
        for i in range(self.input_size):
            genome.node_genes[i] = NodeGene(i, 'input')

        # Output nodes (IDs start after input nodes)
        for i in range(self.output_size):
            genome.node_genes[self.input_size + i] = NodeGene(self.input_size + i, 'output')

        # Initial connections (fully connected input to output)
        for i_in in range(self.input_size):
            for i_out in range(self.output_size):
                in_node_id = i_in
                out_node_id = self.input_size + i_out
                weight = np.random.uniform(-1, 1)
                innovation_num = self.innovation_tracker.get_innovation_num(in_node_id, out_node_id)
                genome.connection_genes[innovation_num] = ConnectionGene(in_node_id, out_node_id, weight, innovation_num)
        return genome

    def speciate_population(self):
        for s in self.species:
            s.genomes = []  # Clear previous species assignments

        for genome in self.population:
            found_species = False
            for s in self.species:
                if self.get_compatibility_distance(genome, s.representative) < COMPATIBILITY_THRESHOLD:
                    s.add_genome(genome)
                    found_species = True
                    break
            if not found_species:
                new_species = Species(self.next_species_id, genome)
                self.next_species_id += 1
                new_species.add_genome(genome)
                self.species.append(new_species)

        # Remove empty species
        self.species = [s for s in self.species if s.genomes]

        # Update representatives for next generation
        for s in self.species:
            s.select_representative()


    def get_compatibility_distance(self, g1, g2):
        # Calculate compatibility distance between two genomes
        # Based on excess, disjoint, and average weight differences
        # Implementation is non-trivial and requires careful gene comparison
        # (matching innovation numbers)

        # Simplified: Count excess/disjoint and average weight difference
        # This is a placeholder for a more robust implementation

        g1_conn_innovs = set(g1.connection_genes.keys())
        g2_conn_innovs = set(g2.connection_genes.keys())

        all_innovs = g1_conn_innovs.union(g2_conn_innovs)

        excess_genes = 0
        disjoint_genes = 0
        weight_diff_sum = 0
        matching_genes_count = 0

        max_g1_innov = max(g1_conn_innovs) if g1_conn_innovs else 0
        max_g2_innov = max(g2_conn_innovs) if g2_conn_innovs else 0

        for innov_num in all_innovs:
            in_g1 = innov_num in g1_conn_innovs
            in_g2 = innov_num in g2_conn_innovs

            if in_g1 and in_g2:
                weight_diff_sum += abs(g1.connection_genes[innov_num].weight - g2.connection_genes[innov_num].weight)
                matching_genes_count += 1
            elif in_g1 and not in_g2:
                if innov_num > max_g2_innov: # Excess
                    excess_genes += 1
                else: # Disjoint
                    disjoint_genes += 1
            elif not in_g1 and in_g2:
                if innov_num > max_g1_innov: # Excess
                    excess_genes += 1
                else: # Disjoint
                    disjoint_genes += 1

        N = max(len(g1.connection_genes), len(g2.connection_genes)) # Normalization factor
        if N < 20: N = 1 # Avoid division by zero/small N issues

        distance = (C1 * excess_genes / N) + (C2 * disjoint_genes / N)
        if matching_genes_count > 0:
            distance += (C3 * weight_diff_sum / matching_genes_count)

        return distance


    def reproduce(self):
        new_population = []
        total_adjusted_fitness = sum(s.average_fitness for s in self.species)

        for species in self.species:
            if not species.genomes: continue

            # Sort genomes by fitness within the species
            species.sort_genomes()

            # Elitism: Keep a percentage of the fittest genomes from each species
            num_elite = int(ELITISM_PERCENTAGE * len(species.genomes))
            new_population.extend(species.genomes[:num_elite])

            # Calculate number of offspring for this species
            if total_adjusted_fitness > 0:
                num_offspring = int(round((species.average_fitness / total_adjusted_fitness) * (POPULATION_SIZE - len(new_population))))
            else:
                num_offspring = 0 # No fitness, no offspring

            # Ensure at least one offspring if species has genomes and isn't elite
            if num_offspring == 0 and len(species.genomes) > 0 and num_elite == 0:
                num_offspring = 1 # At least one offspring for viable species

            for _ in range(num_offspring):
                parent1 = random.choice(species.genomes)
                # 75% chance of crossover with another member of the species
                if random.random() < 0.75 and len(species.genomes) > 1:
                    parent2 = random.choice(species.genomes)
                    child = self.crossover(parent1, parent2)
                else: # Asexual reproduction (mutation only)
                    child = copy.deepcopy(parent1)

                self.mutate(child)
                child.genome_id = self.next_genome_id
                self.next_genome_id += 1
                new_population.append(child)

        # If population size is less than desired, add more mutated elites or random new genomes
        while len(new_population) < POPULATION_SIZE:
            # Pick a random elite or a random genome from the previous population
            if self.population:
                genome_to_copy = random.choice(self.population)
                new_genome = copy.deepcopy(genome_to_copy)
                self.mutate(new_genome)
                new_genome.genome_id = self.next_genome_id
                self.next_genome_id += 1
                new_population.append(new_genome)
            else: # Should not happen after initial population
                new_population.append(self._create_minimal_genome())
                self.next_genome_id += 1

        self.population = new_population[:POPULATION_SIZE] # Trim if too many

    def crossover(self, g1, g2):
        # Assumes g1 is the fitter parent if fitnesses differ
        if g1.fitness < g2.fitness:
            g1, g2 = g2, g1 # Ensure g1 is the fitter parent

        child = Genome(-1) # ID will be assigned later

        # Inherit node genes
        all_node_ids = set(g1.node_genes.keys()).union(g2.node_genes.keys())
        for node_id in all_node_ids:
            if node_id in g1.node_genes and node_id in g2.node_genes:
                child.node_genes[node_id] = copy.deepcopy(random.choice([g1.node_genes[node_id], g2.node_genes[node_id]]))
            elif node_id in g1.node_genes:
                child.node_genes[node_id] = copy.deepcopy(g1.node_genes[node_id])
            elif node_id in g2.node_genes: # This only happens if g2 has excess/disjoint node that g1 doesn't.
                                           # For crossover, usually only fitter parent's disjoint/excess are inherited.
                                           # But for full NEAT, it's more complex with innovation numbers.
                pass # For simplicity, only inherit from fitter parent's excess/disjoint nodes


        # Inherit connection genes
        # For matching genes, pick one randomly.
        # For disjoint/excess genes, only inherit from the fitter parent (g1).
        for innov_num, conn1 in g1.connection_genes.items():
            if innov_num in g2.connection_genes:
                # Matching gene: inherit randomly
                conn2 = g2.connection_genes[innov_num]
                child.connection_genes[innov_num] = copy.deepcopy(random.choice([conn1, conn2]))
                # If one is disabled and the other enabled, 75% chance to inherit disabled
                if not conn1.enabled or not conn2.enabled:
                    if random.random() < 0.75:
                        child.connection_genes[innov_num].enabled = False
                    else:
                        child.connection_genes[innov_num].enabled = True # Enable if both were disabled and 25% chance
            else:
                # Disjoint or excess gene in fitter parent (g1): inherit
                child.connection_genes[innov_num] = copy.deepcopy(conn1)

        return child

    def mutate(self, genome):
        # Mutate weights
        for conn in genome.connection_genes.values():
            if random.random() < MUTATION_RATE_PERTURB_WEIGHT:
                conn.weight += np.random.normal(0, WEIGHT_PERTURB_SCALE)
            elif random.random() < MUTATION_RATE_RESET_WEIGHT:
                conn.weight = np.random.uniform(-1, 1)

        for node in genome.node_genes.values():
            node.bias += np.random.normal(0, WEIGHT_PERTURB_SCALE) # Mutate bias

        # Add connection mutation
        if random.random() < MUTATION_RATE_ADD_CONNECTION:
            self._mutate_add_connection(genome)

        # Add node mutation
        if random.random() < MUTATION_RATE_ADD_NODE:
            self._mutate_add_node(genome)

    def _mutate_add_connection(self, genome):
        available_in_nodes = [node_id for node_id, node_gene in genome.node_genes.items() if node_gene.node_type != 'output']
        available_out_nodes = [node_id for node_id, node_gene in genome.node_genes.items() if node_gene.node_type != 'input']

        if not available_in_nodes or not available_out_nodes:
            return # Cannot add connection

        in_node_id = random.choice(available_in_nodes)
        out_node_id = random.choice(available_out_nodes)

        # Avoid self-loops and existing connections (for feed-forward)
        if in_node_id == out_node_id:
            return

        # Check if connection already exists
        for conn in genome.connection_genes.values():
            if conn.in_node_id == in_node_id and conn.out_node_id == out_node_id:
                return # Connection already exists

        # Basic check for cycles (simplified for feed-forward, full check is complex)
        # For strictly feed-forward, ensure out_node_id's level is higher than in_node_id's.
        # This requires a proper graph representation and topological sorting.
        # For a basic implementation, we might just assume it's feed-forward
        # if in_node_id is an input and out_node_id is an output/hidden or vice versa.
        # A full topological sort is needed for robust cycle detection.

        weight = np.random.uniform(-1, 1)
        innovation_num = self.innovation_tracker.get_innovation_num(in_node_id, out_node_id)
        genome.connection_genes[innovation_num] = ConnectionGene(in_node_id, out_node_id, weight, innovation_num)


    def _mutate_add_node(self, genome):
        # Choose an existing enabled connection to split
        enabled_connections = [conn for conn in genome.connection_genes.values() if conn.enabled]
        if not enabled_connections:
            return # No connections to split

        old_connection = random.choice(enabled_connections)
        old_connection.enabled = False # Disable the old connection

        new_node_id = self.innovation_tracker.get_next_node_id()
        genome.node_genes[new_node_id] = NodeGene(new_node_id, 'hidden')

        # Add two new connections: in_node -> new_node and new_node -> out_node
        # First connection with weight 1.0 (or old_connection.weight)
        # Second connection with old_connection.weight (or 1.0)

        # First new connection
        innov_num1 = self.innovation_tracker.get_innovation_num(old_connection.in_node_id, new_node_id)
        genome.connection_genes[innov_num1] = ConnectionGene(old_connection.in_node_id, new_node_id, 1.0, innov_num1)

        # Second new connection
        innov_num2 = self.innovation_tracker.get_innovation_num(new_node_id, old_connection.out_node_id)
        genome.connection_genes[innov_num2] = ConnectionGene(new_node_id, old_connection.out_node_id, old_connection.weight, innov_num2)


    def evaluate_population(self, env, num_games_per_genome=5):
        for genome in self.population:
            total_score = 0
            for _ in range(num_games_per_genome):
                obs = env.reset() # SlimeVolley-v0 reset returns obs, info
                done = False
                game_score = 0
                while not done:
                    # Agent's action
                    agent_obs = np.array(obs[6:]) # Agent's own state (last 6 elements)
                    ball_obs = np.array(obs[:6]) # Ball state (first 6 elements)

                    # Ensure input matches what the network expects (e.g., ball_x, ball_y, ball_vx, ball_vy, agent_x, agent_y, agent_vx, agent_vy, opponent_x, opponent_y, opponent_vx, opponent_vy)
                    # SlimeVolley-v0 obs: [ball.x, ball.y, ball.vx, ball.vy,
                    #                      player.x, player.y, player.vx, player.vy,
                    #                      opponent.x, opponent.y, opponent.vx, opponent.vy]
                    # Let's directly use the 12-dim observation for simplicity for the NEAT agent.
                    nn_input = obs

                    output = genome.calculate_output(nn_input)
                    action = (output > 0.5).astype(int) # Convert to binary actions

                    # SlimeVolley built-in opponent (index 0 for left player)
                    # The environment internally handles the opponent.
                    # We are training our agent (player 1, on the right) against the built-in AI.
                    # The `env.step` takes a single action for the agent.
                    # The `info` dict may contain 'otherObs' if you want to get opponent's observation.

                    obs, reward, done,info = env.step(action)
                    game_score += reward
                    if done:
                        break
                total_score += game_score
            # print(f"Genome :{genome.genome_id} Total score  Score: {total_score}")

            genome.fitness = total_score / num_games_per_genome # Average score

    def run(self, env):
        self.create_initial_population()

        for gen in range(NUM_GENERATIONS):
            self.generation = gen
            print(f"Generation {gen+1}")

            self.speciate_population()
            print(f"  Number of species: {len(self.species)}")

            # Evaluate fitness of each genome
            self.evaluate_population(env,10)

            # Calculate adjusted fitness for speciation
            for species in self.species:
                species.calculate_average_fitness()

            # Find best genome
            best_genome = None
            max_fitness = -np.inf
            for genome in self.population:
                if genome.fitness > max_fitness:
                    max_fitness = genome.fitness
                    best_genome = genome

            print(f"  Max fitness: {max_fitness:.2f}")

            if max_fitness >= 100.0: # If we can beat the AI consistently (average score 1.0 means always winning)
                print(f"  AI beaten at generation {gen+1}!")
                return best_genome

            self.reproduce()

        return best_genome # Return the best performing genome after all generations
```

```python


# It's good practice to ensure the SlimeVolley-v0 environment is registered


class SurvivalRewardEnv(gym.RewardWrapper):
  """
  A RewardWrapper for Gymnasium environments that adds a small
  survival bonus to the reward at each timestep.

  This encourages the agent to prolong the episode.
  """
  def __init__(self, env):
    """
    Initializes the SurvivalRewardEnv wrapper.

    :param env: (Gymnasium Environment) The environment to wrap.
    """
    super().__init__(env) # Preferred way to call parent constructor in Python 3+
    print("SurvivalRewardEnv initialized: Adding +0.01 reward per timestep.")

  def reward(self, reward):
    """
    Modifies the reward by adding a survival bonus.

    :param reward: (float) The original reward from the wrapped environment.
    :return: (float) The modified reward.
    """
    # Add a small positive constant to the reward for each timestep
    return reward + 0.001





```
