```python
import argparse
import os
import shutil
import jax
from evojax.task.slimevolley import SlimeVolley
from evojax.policy.mlp import MLPPolicy
from evojax.algo import CMA
from evojax import Trainer
from evojax import util
```

```python
# Configuration
class Config:
    def __init__(self):
        self.pop_size = 128
        self.hidden_size = 20
        self.num_tests = 100
        self.n_repeats = 16
        self.max_iter = 500
        self.test_interval = 50
        self.log_interval = 10
        self.seed = 123
        self.init_std = 0.5
        self.debug = False

config = Config()
print("Setup completed!")
```

```python
# Create log directory
log_dir = './log/slimevolley'
if not os.path.exists(log_dir):
    os.makedirs(log_dir, exist_ok=True)

# Setup logger
logger = util.create_logger(
    name='SlimeVolley', log_dir=log_dir, debug=config.debug)

logger.info('EvoJAX SlimeVolley')
logger.info('=' * 30)

# Initialize environment and policy
max_steps = 3000
train_task = SlimeVolley(test=False, max_steps=max_steps)
test_task = SlimeVolley(test=True, max_steps=max_steps)

print(f"Observation shape: {train_task.obs_shape}")
print(f"Action shape: {train_task.act_shape}")

```

```python
# Create policy network
policy = MLPPolicy(
    input_dim=train_task.obs_shape[0],
    hidden_dims=[config.hidden_size],
    output_dim=train_task.act_shape[0],
    output_act_fn='tanh',
)

print(f"Policy parameters: {policy.num_params}")


```

```python
# Create evolutionary solver
solver = CMA(
    pop_size=config.pop_size,
    param_size=policy.num_params,
    init_stdev=config.init_std,
    seed=config.seed,
    logger=logger,
)
print("Environment and components initialized!")
```

#Training


```python
# Create trainer
trainer = Trainer(
    policy=policy,
    solver=solver,
    train_task=train_task,
    test_task=test_task,
    max_iter=config.max_iter,
    log_interval=config.log_interval,
    test_interval=config.test_interval,
    n_repeats=config.n_repeats,
    n_evaluations=config.num_tests,
    seed=config.seed,
    log_dir=log_dir,
    logger=logger,
)

print("Starting training...")
print(f"Population size: {config.pop_size}")
print(f"Max iterations: {config.max_iter}")
print(f"Hidden layer size: {config.hidden_size}")
print("-" * 50)

# Run training
trainer.run(demo_mode=False)

print("Training completed!")
```

#Testing


```python
# Copy best model for testing
src_file = os.path.join(log_dir, 'best.npz')
tar_file = os.path.join(log_dir, 'model.npz')

if os.path.exists(src_file):
    shutil.copy(src_file, tar_file)
    trainer.model_dir = log_dir

    print("Testing the trained model...")
    print("-" * 30)

    # Run test
    trainer.run(demo_mode=True)

    print("Testing completed!")
else:
    print("No trained model found. Please run training first.")
```

```python
# Plot training progress if log files exist
import glob

def plot_training_progress():
    # Look for training log files
    log_files = glob.glob(os.path.join(log_dir, "*.log"))

    if not log_files:
        print("No log files found for plotting.")
        return

    # Try to extract fitness values from logs
    try:
        fitness_values = []
        iterations = []

        with open(log_files[0], 'r') as f:
            lines = f.readlines()

        for line in lines:
            if 'Iter' in line and 'fitness' in line:
                # Parse iteration and fitness from log line
                parts = line.split()
                for i, part in enumerate(parts):
                    if part == 'Iter':
                        iter_num = int(parts[i+1])
                        iterations.append(iter_num)
                    elif 'fitness' in part:
                        # Find fitness value
                        for j in range(i, len(parts)):
                            try:
                                fitness = float(parts[j])
                                fitness_values.append(fitness)
                                break
                            except ValueError:
                                continue
                        break

        if fitness_values and iterations:
            plt.figure(figsize=(12, 6))

            plt.subplot(1, 2, 1)
            plt.plot(iterations, fitness_values, 'b-', linewidth=2)
            plt.xlabel('Iteration')
            plt.ylabel('Fitness')
            plt.title('Training Progress')
            plt.grid(True, alpha=0.3)

            plt.subplot(1, 2, 2)
            # Plot last 100 iterations for detail
            if len(fitness_values) > 100:
                recent_iters = iterations[-100:]
                recent_fitness = fitness_values[-100:]
                plt.plot(recent_iters, recent_fitness, 'r-', linewidth=2)
                plt.xlabel('Iteration')
                plt.ylabel('Fitness')
                plt.title('Recent Progress (Last 100 iterations)')
            else:
                plt.plot(iterations, fitness_values, 'r-', linewidth=2)
                plt.xlabel('Iteration')
                plt.ylabel('Fitness')
                plt.title('Training Progress')
            plt.grid(True, alpha=0.3)

            plt.tight_layout()
            plt.show()

            print(f"Best fitness achieved: {max(fitness_values):.3f}")
        else:
            print("Could not extract fitness values from logs.")

    except Exception as e:
        print(f"Error plotting training progress: {e}")

plot_training_progress()
```

```python
# Analyze the trained agent's performance
def analyze_performance():
    try:
        print("Performance Analysis")
        print("=" * 40)

        # Load best parameters
        best_params_file = os.path.join(log_dir, 'best.npz')
        if not os.path.exists(best_params_file):
            print("No trained model found.")
            return

        # Run multiple test episodes
        task_reset_fn = jax.jit(test_task.reset)
        policy_reset_fn = jax.jit(policy.reset)
        step_fn = jax.jit(test_task.step)
        action_fn = jax.jit(policy.get_actions)

        best_params = trainer.solver.best_params[None, :]

        episode_rewards = []
        episode_lengths = []
        wins = 0

        num_test_episodes = 10
        print(f"Running {num_test_episodes} test episodes...")

        for episode in range(num_test_episodes):
            key = jax.random.PRNGKey(episode)[None, :]
            task_state = task_reset_fn(key)
            policy_state = policy_reset_fn(task_state)

            total_reward = 0

            for step in range(max_steps):
                action, policy_state = action_fn(task_state, best_params, policy_state)
                task_state, reward, done = step_fn(task_state, action)
                total_reward += float(reward[0])

                if done[0]:
                    break

            episode_rewards.append(total_reward)
            episode_lengths.append(step + 1)

            if total_reward > 0:
                wins += 1

            print(f"Episode {episode + 1}: Reward = {total_reward:.2f}, Length = {step + 1}")

        # Statistics
        print("\nPerformance Statistics:")
        print(f"Average reward: {np.mean(episode_rewards):.2f} ± {np.std(episode_rewards):.2f}")
        print(f"Best episode reward: {max(episode_rewards):.2f}")
        print(f"Worst episode reward: {min(episode_rewards):.2f}")
        print(f"Average episode length: {np.mean(episode_lengths):.1f}")
        print(f"Win rate: {wins}/{num_test_episodes} ({100*wins/num_test_episodes:.1f}%)")

        # Plot episode rewards
        plt.figure(figsize=(10, 4))

        plt.subplot(1, 2, 1)
        plt.bar(range(1, num_test_episodes + 1), episode_rewards)
        plt.xlabel('Episode')
        plt.ylabel('Total Reward')
        plt.title('Test Episode Rewards')
        plt.axhline(y=0, color='r', linestyle='--', alpha=0.5)
        plt.grid(True, alpha=0.3)

        plt.subplot(1, 2, 2)
        plt.hist(episode_rewards, bins=5, alpha=0.7, edgecolor='black')
        plt.xlabel('Episode Reward')
        plt.ylabel('Frequency')
        plt.title('Reward Distribution')
        plt.axvline(x=0, color='r', linestyle='--', alpha=0.5)
        plt.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

    except Exception as e:
        print(f"Error in performance analysis: {e}")

analyze_performance()
```

```python
# Functions to save and load the trained model
def save_model(filename='slimevolley_model.npz'):
    """Save the trained model parameters."""
    try:
        best_params_file = os.path.join(log_dir, 'best.npz')
        if os.path.exists(best_params_file):
            shutil.copy(best_params_file, filename)
            print(f"Model saved as {filename}")
        else:
            print("No trained model found to save.")
    except Exception as e:
        print(f"Error saving model: {e}")

def load_model(filename='slimevolley_model.npz'):
    """Load a saved model."""
    try:
        if os.path.exists(filename):
            shutil.copy(filename, os.path.join(log_dir, 'model.npz'))
            print(f"Model loaded from {filename}")
        else:
            print(f"File {filename} not found.")
    except Exception as e:
        print(f"Error loading model: {e}")

# Save the current best model
save_model()

print("Model management functions ready!")
print("Use save_model('filename.npz') to save")
print("Use load_model('filename.npz') to load")
```
